# Slicing

```python
Fruits_list = ["mango","banana","apple","apple","grape","passion","pineapple"]
```

```python
#starting point: the end point (exclusive)
Fruits_list[2 : 5]
```

```python
Fruits_list[4:]
```

```python
Fruits_list[:-5]
```

```python
import numpy as np
A = np.array([[2,1],[1,2]])
eigenvalues, eigenvectors = np.linalg.eig(A)
Lambda = np.diag(eigenvalues)
# Q contains the eigenvectors
Q = eigenvectors
Q_inv = np.linalg.inv(Q)
A_reconstructed = Q @ Lambda @ Q_inv
# Display results print("Original Matrix A:")
print(A)
print("\nEigenvalues:")
print(eigenvalues)
print("\nEigenvectors Q:")
print(Q)
print("\nDiagonal Matrix Lambda:")
print(Lambda)
print("\nReconstructed Matrix:")
print(np.round(A_reconstructed, 6))
```
