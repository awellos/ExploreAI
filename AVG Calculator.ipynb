```python
# program to calculate average of 10 numbers

#initialize sum as 0
sum = 0
#use a for loop to input 10 numbers
for i in range (10):
  a = float(input(f"Enter a{i+1}:"))
#add each number to the sum
  sum += a

 #calculate average of the numbers
  average = sum/10

  print("Average =", average)
```
